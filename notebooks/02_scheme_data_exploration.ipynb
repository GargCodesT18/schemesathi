# 02. Scheme Data Exploration

Explore the curated Indian Government schemes dataset (`data/schemes.json`).
Every scheme includes official provenance, benefits, and deterministic rule criteria.

### 1. Load Schemes Repository

```python
import pandas as pd
from app.data.scheme_repository import SchemeRepository

repo = SchemeRepository()
schemes = repo.get_all()
print(f'Total Curated Schemes: {len(schemes)}')
print(f'Categories: {repo.get_all_categories()}')
```

### 2. Dataset Overview via Pandas DataFrame

```python
df = pd.DataFrame([{
    'id': s.id,
    'short_name': s.short_name or s.name,
    'category': s.category,
    'rules_count': len(s.eligibility_rules),
    'state_scope': ', '.join(s.state_scope),
    'source': s.source_name
} for s in schemes])
df
```

### 3. Distribution by Sector Category

```python
cat_counts = df['category'].value_counts()
print(cat_counts)
```

### 4. Rule Criteria Breakdown
Inspect the deterministic rule format used for eligibility evaluation.

```python
sample_scheme = repo.get_by_id('pm-kisan')
print(f'Scheme: {sample_scheme.name}')
print('Eligibility Rules:')
for r in sample_scheme.eligibility_rules:
    print(f'  • Field: {r.field:15} | Op: {r.operator:20} | Val: {r.value} ({r.description})')
print(f'Official Source URL: {sample_scheme.source_url}')
```

### Interpretation
The dataset contains clean, verified schemes across key citizen sectors with structured criteria ready for deterministic rule evaluation.
