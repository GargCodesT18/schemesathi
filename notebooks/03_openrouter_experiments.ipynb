### 1. Robust JSON Cleaning Utility
Open-weights models often wrap JSON in markdown blocks or commentary. SchemeSathi handles this deterministically.

````python
from app.services.llm_service import LLMService

fenced_input = '''
```json
{
  "age": 42,
  "state": "Punjab",
  "occupation": "farmer",
  "annual_income": 200000
}
```
'''
cleaned = LLMService._clean_json_response(fenced_input)
print('Cleaned JSON:')
print(cleaned)
````

### 2. Safe Parsing with Pydantic Validation

````python
import json
from app.models.profile import CitizenProfile

data = json.loads(cleaned)
profile = CitizenProfile(**data)
print('Parsed CitizenProfile:')
print(profile.model_dump(exclude_none=True))
````
