# 05. Deterministic Eligibility Engine

Demonstrates transparent, criterion-by-criterion rule evaluation in pure Python without LLM hallucination.
Produces four distinct states:
- **LIKELY ELIGIBLE**
- **POSSIBLY ELIGIBLE**
- **MORE INFORMATION NEEDED**
- **NOT ELIGIBLE**

### 1. Initialize Schemes and Eligibility Engine

```python
from app.data.scheme_repository import SchemeRepository
from app.services.eligibility_service import EligibilityEngine
from app.models.profile import CitizenProfile

repo = SchemeRepository()
engine = EligibilityEngine()
schemes = repo.get_all()
print(f'Loaded {len(schemes)} schemes for evaluation.')
```

### 2. Evaluating a Farmer Profile (Scenario 1)

```python
farmer_profile = CitizenProfile(
    age=42,
    state='Punjab',
    occupation='farmer',
    farmer_status=True,
    annual_income=200000,
    rural_or_urban='rural'
)

results = engine.evaluate_all(schemes, farmer_profile)
for res in results[:5]:
    print(f'[{res.status.value}] {res.scheme.name} ({res.scheme.category})')
    for m in res.matched_criteria:
        print(f'   ✓ {m.detail}')
    for u in res.unknown_criteria:
        print(f'   ? {u.detail}')
    for f in res.failed_criteria:
        print(f'   ✗ {f.detail}')
```

### 3. Evaluating an Incomplete Profile (Unknown Handling)

```python
partial_profile = CitizenProfile(occupation='farmer')
res_partial = engine.evaluate_scheme(repo.get_by_id('pm-kmy'), partial_profile)
print(f'Scheme: {res_partial.scheme.name}')
print(f'Status: {res_partial.status.value}')
print(f'Summary: {res_partial.summary}')
print(f'Missing Fields: {res_partial.missing_fields}')
```

### Interpretation
The eligibility engine is fully deterministic, transparent, and auditable. Every decision is tied to an explicit rule.
